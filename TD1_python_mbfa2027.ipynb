```python
import pandas as pd
```

Je crée une table en respectant les données et les types de données imposés. Enfin, j'ajouterai des colonnes supplémentaires pour enrichir mon dataframe.

```python
df['statut_emploi'] = ["CDI", "cdi", "CDD", "indep", "CDI", "c.d.i", "CDI", "CDI"]
df['region'] = ["Nord", "nord", "Sud", "Ouest", "Nord", "Est", None, "Nord"]
df['sexe'] = ["F", "H", "F", "H","H","F","h", "F"]
df['defaut'] = [0, 0, 1, 0, 1, 0, 0, 0]
df['date_ouverture'] = ["05/06/2020", "12/10/2018", "01/09/2019", "30/02/2020", "01/01/2025", "02/05/2035", "17/02/2017", "05/06/2020"]
```

Je constate que certaines données ont été insérées alors qu’elles n’ont pas lieu d’être. J’ai également relevé d’autres incohérences, notamment concernant le dtype de la colonne mais nous procéderons à ces corrections dans quelques instants.

```python
codes = ["ND", "NC", "?", "inconnu"]
df['revenu_mensuel'] = df['revenu_mensuel'].replace(codes, pd.NA) #Je supprime de ma colonne qui sont citées dans la variable code.
```

```python
df['revenu_mensuel'] = pd.to_numeric(
    df['revenu_mensuel']
    .str.replace(r"[€ .]", "", regex=True)
    .str.replace(",", "."),
    errors="coerce")
```

```python
df
```

```python
df['statut_emploi'].unique()
```

Les valeurs dans la colonne *statut_emploi* ne sont pas standardisées.

```python
df['statut_emploi'] = df['statut_emploi'].str.strip().str.lower()
```

```python
df['statut_emploi'] = df['statut_emploi'].replace({
    'c.d.i': 'cdi',
    'c.d.d': 'cdd',
    'indep': 'indépendant'
})
```

```python
df
```

Maintenant, on va procéder à la colonne *age* , où le dtype n'est pas bon ainsi que le "ans" de trop pour l'*id_client* C02..

```python
df['age'] = pd.to_numeric(
    df['age'].str.replace("ans", ""),
    errors="coerce")
```

```python
df
```

```python
df['date_ouverture'] = pd.to_datetime(df['date_ouverture'], format= "%d/%m/%Y", errors= "coerce") #Je met le bon dtype pour la colonne date.
```

```python
df['date_ouverture'].isna().sum()
```

Je vérifie qu'il n'y ait pas d'incohérence sur l'âge, ça doit être compris entre 18 et 100 ans.

```python
a = df['age']
age_invalide = ~a.between(18, 100) & a.notna()
```

```python
df.loc[age_invalide, ["id_client", "age"]] #J'affiche les données en question.
```

```python
df.loc[age_invalide, "age"] = pd.NA #Et je les supprime du jeu de données.
```

Je vérifie qu’il n’y a aucune incohérence dans les taux d’endettement, ceux-ci devant être compris entre 0 et 1.

```python
t = df['taux_endettement']
taux_invalide = ~t.between(0, 1) & t.notna()
```

```python
df.loc[taux_invalide, "taux_endettement"] = pd.NA
```

```python
df
```

```python
df['age'].dropna().between(18, 100).all() #verif
df['taux_endettement'].dropna().between(0, 1).all() #verif
```

Nous allons faire l'écart interquartile.

```python
q1 = r.quantile(0.25)
q3 = r.quantile(0.75)
IQR = q3 - q1
borne_haute = q3 + 1.5 * IQR
```

```python
plafond = r.quantile(0.99)
df['revenu_plaf'] = df['revenu_mensuel'].clip(upper= plafond)
```

Nous allons compléter les données manquantes importantes, puis répartir aléatoirement le jeu de données en deux ensembles : 75 % des données seront utilisées pour entraîner le modèle et les 25 % restantes serviront à le tester.

```python
df_modele = df.drop(columns="revenu_plaf", errors = "ignore").copy()
```

```python
train = df_modele.sample(frac = 0.75, random_state=42).copy()
test = df_modele.drop(index= train.index).copy()
```

```python
train["defaut"].value_counts(normalize=True)
test["defaut"].value_counts(normalize=True)
```

```python
train['revenu_absent'] = train['revenu_mensuel'].isna().astype(int)
test['revenu_absent'] = test['revenu_mensuel'].isna().astype(int)
```

```python
med = train['revenu_mensuel'].median()
train['revenu_mensuel'] = train['revenu_mensuel'].fillna(med)
test['revenu_mensuel'] = test['revenu_mensuel'].fillna(med)
```

```python
train
```

```python
med_age = train["age"].median()
train["age"] = train["age"].fillna(med_age)
test["age"] = test["age"].fillna(med_age)
```

```python
med_taux = train["taux_endettement"].median()
train["taux_endettement"] = train["taux_endettement"].fillna(med_taux)
test["taux_endettement"] = test["taux_endettement"].fillna(med_taux)
```

```python
train
```

```python
mu = train["age"].mean()
sigma = train["age"].std(ddof=0)
train["age_z"] = (train["age"] - mu) / sigma
test["age_z"] = (test["age"] - mu) /sigma
```

```python
train["age_z"].mean().round(3)
```

```python
train["age_z"].std(ddof=0)
```

```python
train["region"] = train["region"].str.strip().str.lower()
test["region"] = test["region"].str.strip().str.lower()
```

Je devais le faire car pour *C02*, il y avait "nord" alors que les autres étaient uniformes. Je devais harmoniser.

Pour que le modèle soit plus facilement entraînable, je dois faire un OneHot Encoding pour *Region* et *Statut_Emploi*.

```python
cat = ["region", "statut_emploi"]
train = pd.get_dummies(train, columns=cat, dtype=int)
test = pd.get_dummies(test, columns=cat, dtype=int)
```

```python
train
```

Je dois réindexer la table.

```python
test = test.reindex(columns=train.columns, fill_value=0)
train.columns.equals(test.columns)
```

```python
train["sexe"] = train["sexe"].astype("string").str.strip().str.upper()
test["sexe"] = test["sexe"].astype("string").str.strip().str.upper()
```

```python
train["sexe_H"] = train["sexe"].map({"F": 0, "H":1})
test["sexe_H"] = test["sexe"].map({"F": 0, "H": 1})
```
